```python
from fafbseg import flywire
import json
```

```python
# DEFENSE #1 - Manually find the synapses between two adjacent neurons as per the Stanley paper.
# Using get_connectivity() rather than get_adjacency() as that's what the main code is using.

neurons_by_group = None
with open('/Users/ayushshrivastava/code/lab-analysis-test/connectomics-tests/pathways/data/groups.json') as f:
    neurons_by_group = json.load(f)
```

```python
new_ovidn_ids = flywire.update_ids(neurons_by_group['OviDN'])
print(new_ovidn_ids[new_ovidn_ids['changed'] == True].head())
new_ovidn_ids = new_ovidn_ids['new_id']
new_glu_ids = flywire.update_ids(neurons_by_group['Stanley Glu interneuron'])
print(new_glu_ids[new_glu_ids['changed'] == True].head())
new_glu_ids = new_glu_ids['new_id']
```

```python
# Grab Ir94e neurons

ir94e = flywire.synapses.get_connectivity(neurons_by_group['Ir94e'], upstream=False)
ir94e = ir94e[ir94e['weight'] >= 5]
```

```python
# Test 1 - Ir94e -> X ✅
print("Ir94e -> GNG.SLP.T1 (L):", ir94e[ir94e['post'] == neurons_by_group['GNG.SLP.T1 (L)'][0]].sum()['weight'])
print("Ir94e -> GNG.SLP.T1 (R):", ir94e[ir94e['post'] == neurons_by_group['GNG.SLP.T1 (R)'][0]].sum()['weight'])
print("Ir94e -> GNG.SLP.T2 (L):", ir94e[ir94e['post'] == neurons_by_group['GNG.SLP.T2 (L)'][0]].sum()['weight'])
print("Ir94e -> GNG.SLP.T2 (R):", ir94e[ir94e['post'] == neurons_by_group['GNG.SLP.T2 (R)'][0]].sum()['weight'])
print("Ir94e -> Earmuff:", ir94e[ir94e['post'] == neurons_by_group['Earmuff'][0]].sum()['weight'])
```

```python
gng_t1_L = flywire.synapses.get_connectivity(neurons_by_group['GNG.SLP.T1 (L)'], upstream=False)
gng_t1_L = gng_t1_L[gng_t1_L['weight'] >= 5]
```

```python
# Test 2 - GNG.SLP.T1 (L) -> X ❌
print("GNG.SLP.T1 (L) -> Stanley ACh:", gng_t1_L[gng_t1_L['post'].isin(neurons_by_group['Stanley ACh interneuron'])].sum()['weight'])
print("GNG.SLP.T1 (L) -> Stanley Glu:", gng_t1_L[gng_t1_L['post'].isin(new_glu_ids)].sum()['weight']) # Undercounted (5 vs 13)
```

```python
gng_t1_R = flywire.synapses.get_connectivity(neurons_by_group['GNG.SLP.T1 (R)'], upstream=False)
gng_t1_R = gng_t1_R[gng_t1_R['weight'] >= 5]
```

```python
# Test 3 - GNG.SLP.T1 (R) -> X ✅
print("GNG.SLP.T1 (R) -> Stanley ACh:", gng_t1_R[gng_t1_R['post'].isin(neurons_by_group['Stanley ACh interneuron'])].sum()['weight'])
print("GNG.SLP.T1 (R) -> Stanley Glu:", gng_t1_R[gng_t1_R['post'].isin(new_glu_ids)].sum()['weight'])
```

```python
gng_t2_L = flywire.synapses.get_connectivity(neurons_by_group['GNG.SLP.T2 (L)'], upstream=False)
gng_t2_L = gng_t2_L[gng_t2_L['weight'] >= 5]
```

```python
# Test 4 - GNG.SLP.T2 (L) -> X ✅
print("GNG.SLP.T2 (L) -> Stanley ACh:", gng_t2_L[gng_t2_L['post'].isin(neurons_by_group['Stanley ACh interneuron'])].sum()['weight'])
print("GNG.SLP.T2 (L) -> Stanley Glu:", gng_t2_L[gng_t2_L['post'].isin(new_glu_ids)].sum()['weight'])
```

```python
gng_t2_R = flywire.synapses.get_connectivity(neurons_by_group['GNG.SLP.T2 (R)'], upstream=False)
gng_t2_R = gng_t2_R[gng_t2_R['weight'] >= 5]
```

```python
# Test 5 - GNG.SLP.T2 (R) -> X ⚠️
print("GNG.SLP.T2 (R) -> Stanley ACh:", gng_t2_R[gng_t2_R['post'].isin(neurons_by_group['Stanley ACh interneuron'])].sum()['weight'])
print("GNG.SLP.T2 (R) -> Stanley Glu:", gng_t2_R[gng_t2_R['post'].isin(new_glu_ids)].sum()['weight']) # Undercounted (24 vs 25)
```

```python
earmuff = flywire.synapses.get_connectivity(neurons_by_group['Earmuff'][0], upstream=False)
earmuff = earmuff[earmuff['weight'] >= 5]
```

```python
# Test 6 - Earmuff -> X ⚠️
print("Earmuff -> OviDN:", earmuff[earmuff['post'].isin(new_ovidn_ids)].sum()['weight']) # Overcounted by 4
print("Earmuff -> Stanley ACh:", earmuff[earmuff['post'].isin(neurons_by_group['Stanley ACh interneuron'])].sum()['weight'])
print("Earmuff -> Stanley Glu:", earmuff[earmuff['post'].isin(new_glu_ids)].sum()['weight'])
```

```python
# Test 7 - Stanley ACh -> OviDN ⚠️
ach = flywire.synapses.get_connectivity(neurons_by_group["Stanley ACh interneuron"], upstream=False)
ach = ach[ach['weight'] >= 5]
print("Stanley ACh -> OviDN", ach[ach['post'].isin(new_ovidn_ids)].sum()['weight']) # Overcounted by 2
```

```python
# Test 8 - Stanley Glu -> OviDN ⚠️
print(new_glu_ids)
glu = flywire.synapses.get_connectivity(new_glu_ids, upstream=False, materialization=783)
glu = glu[glu['weight'] >= 5]
print("Stanley Glu -> OviDN", glu[glu['post'].isin(new_ovidn_ids)].sum()['weight']) # Overcounted by 1 (32 vs 31)
```

```python
# Test 9 - OviDN => X ⚠️
ovidn = flywire.synapses.get_connectivity(new_ovidn_ids, downstream=False, materialization=783)
ovidn = ovidn[ovidn['weight'] >= 5]
print("OviDN => Stanley Glu", ovidn[ovidn['pre'].isin(new_glu_ids)].sum()['weight']) # Overcounted by 1 (32 vs 31)
print("OviDN => Stanley ACh", ovidn[ovidn['pre'].isin(neurons_by_group["Stanley ACh interneuron"])].sum()['weight']) # Overcounted by 2 (112 vs 110)
print("OviDN => Earmuff", ovidn[ovidn['pre'].isin(neurons_by_group["Earmuff"])].sum()['weight']) # Overcounted by 4 (87 vs 83)
```
